## Task-8: Calculate the average investment duration.


```python
import pandas as pd

from google.colab import files
uploads = files.upload()
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
df.head()
```

```python
# Explore the "Duration" column
print("Unique Values in the 'Duration' Column:")
print(df["Duration"].unique())
```

```python
# Map durations to numeric values
duration_mapping = {
    "Less than 1 year": 0.5,
    "1-3 years": 2,
    "3-5 years": 4,
    "More than 5 years": 6
}

# Creating a new column with numeric values
df["Numeric_Duration"] = df["Duration"].map(duration_mapping)

# Check the new column
print(df[["Duration", "Numeric_Duration"]].head())
```

```python
# Calculate the average investment duration
average_duration = df["Numeric_Duration"].mean()

# Result
print(f"The average investment duration is approximately {average_duration:.2f} years.")
```

## Task-6: Identify and present main savings objectives.

```python
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
uploads = files.upload()
df = pd.read_csv("Data_set 2 - Copy.csv")

# Analyzing savings objective
savings_objectives = df["What are your savings objectives?"].value_counts()

print("Savings Objectives Summary:")
print(savings_objectives)
```

```python
# Save the summary into a DataFrame
savings_df = savings_objectives.reset_index()
savings_df.columns = ["Savings Objective", "Frequency"]
print(savings_df)
```

```python
plt.figure(figsize=(8, 6))
plt.bar(savings_df["Savings Objective"], savings_df["Frequency"], color="skyblue")
plt.title("Savings Objectives Distribution")
plt.xlabel("Savings Objectives")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# So we can understand from this Many people have savings Objectives for Retirement Plan

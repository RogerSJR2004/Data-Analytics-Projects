#Task-7 : Common Information Sources

```python
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt

from google.colab import files
uploads = files.upload()
df = pd.read_csv("Data_set 2 - Copy.csv")

# View the first few rows
print(df["Source"].head())

# Check for missing
print(df["Source"].isnull().sum())
```

```python
sources = df["Source"].dropna().tolist()  # Removing any null values

# Count the occurrences of each source
source_counts = Counter(sources)

# Display the summary of common sources
print("Summary of Common Information Sources:")
for source, count in source_counts.items():
    print(f"{source}: {count} participants")

# Convert to a DataFrame
source_df = pd.DataFrame(source_counts.items(), columns=["Source", "Frequency"])
source_df = source_df.sort_values(by="Frequency", ascending=False)
```

```python
plt.figure(figsize=(8, 6))
plt.bar(source_df["Source"], source_df["Frequency"], color="yellow")
plt.title("Common Sources of Investment Information")
plt.xlabel("Source")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

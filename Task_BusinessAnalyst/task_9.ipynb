##Task-9: Participants expectations from investments

```python
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt

from google.colab import files
uploads = files.upload()
df = pd.read_csv("Data_set 2 - Copy.csv")
df
```

```python
# Explore the 'Expect' column
print("Unique Values in the 'Expect' Column:")
print(df["Expect"].unique())

# finding the missing values
print("Missing Values in 'Expect' Column:")
print(df["Expect"].isnull().sum())
```

```python
# Drop missing values
expectations = df["Expect"].dropna().tolist()

# Count the occurrences
expect_counts = Counter(expectations)

# print summary
print("Summary of Expectations:")
for expectation, count in expect_counts.items():
    print(f"{expectation}: {count} participants")

# Convert to DataFrame
expect_df = pd.DataFrame(expect_counts.items(), columns=["Expectation", "Frequency"])
expect_df = expect_df.sort_values(by="Frequency", ascending=False)
```

```python
# viuals for better understanding
plt.figure(figsize=(8, 6))
plt.bar(expect_df["Expectation"], expect_df["Frequency"], color="skyblue")
plt.title("Common Investment Expectations")
plt.xlabel("Expectation")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
